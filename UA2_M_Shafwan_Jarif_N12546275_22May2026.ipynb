## Cleaning Structured Data:
The structured datasets contain additional formatting rows and metadata that are not required for the analysis. Therefore, the data is cleaned to improve consistency and ensure that only meaningful participation information is included. Cleaning the datasets involves removing completely empty rows, selecting relevant columns, and standardising column names where necessary. ## Final Data Cleaning and Preparation:
After successfully loading the datasets, additional cleaning is performed to improve consistency and prepare the data for analysis. The first column currently appears as “Unnamed: 0”, which contains the names of sporting activities. This column is renamed to “Activity” to improve readability and make the dataset easier to work with during filtering and visualisation. The datasets also contain some remaining formatting rows that are not actual sports participation records, including the “Participation rate” row and empty values. These rows are removed so that only valid sports participation data remains. Finally, the participation columns are converted into numeric format to ensure that calculations, comparisons, and visualisations can be performed accurately throughout the analysis.

```python
sport_children = sport_children.rename(columns={
    'Unnamed: 0': 'Activity'
})

qld_children = qld_children.rename(columns={
    'Unnamed: 0': 'Activity'
})


sport_children = sport_children[
    sport_children['Activity'].notna()
]

qld_children = qld_children[
    qld_children['Activity'].notna()
]


sport_children = sport_children[
    sport_children['Activity'] != 'Participation rate'
]

qld_children = qld_children[
    qld_children['Activity'] != 'Participation rate'
]


sport_children = sport_children[
    ['Activity', 'Total#', 'Males', 'Females', '0-4', '5-8', '9-11', '12-14']
]

qld_children = qld_children[
    ['Activity', 'Total#', 'Males', 'Females', '0-4', '5-8', '9-11', '12-14']
]


numeric_columns = [
    'Total#',
    'Males',
    'Females',
    '0-4',
    '5-8',
    '9-11',
    '12-14'
]

for col in numeric_columns:
    sport_children[col] = pd.to_numeric(
        sport_children[col],
        errors='coerce'
    )

    qld_children[col] = pd.to_numeric(
        qld_children[col],
        errors='coerce'
    )


sport_children.head()
```

How can participation patterns across different youth age groups be analysed to identify distinct Olympic sport engagement pathways that may support long-term participation, athlete development, and future fan engagement leading into the Brisbane 2032 Olympic Games?

Understanding how participation changes across childhood and adolescent age groups is important because Olympic sports often rely on long-term athlete development pathways that begin at an early age. Some sports may maintain strong participation consistently as children grow older, while other sports may experience declining participation during adolescence. This analysis extends beyond basic descriptive statistics by applying both correlation analysis and machine learning techniques to youth participation data. Using machine learning helps group sports with similar engagement characteristics together, allowing deeper strategic interpretation beyond simple participation comparisons.

The analysis may help identify:

- sports with strong long-term youth retention
- sports that attract participants later in childhood

```python
age_columns = ['0-4', '5-8', '9-11', '12-14']
age_data = sport_filtered[['Activity'] + age_columns].copy()
```

```python
for col in age_columns:
    age_data[col] = pd.to_numeric(age_data[col], errors='coerce')

age_data
```

## Correlation Analysis:
Correlation analysis is used to examine how strongly participation rates between different age groups are related. Strong positive correlations may suggest that sports maintain participation consistently as children grow older, while weaker relationships may indicate participation decline or inconsistent retention pathways. Understanding these relationships may help identify which sports currently support stronger long-term engagement opportunities relevant to future Olympic participation and fan development.

```python
correlation_matrix = age_data[age_columns].corr()
correlation_matrix
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='Blues',
    linewidths=0.5
)

plt.title('Correlation Between Youth Age Group Participation Rates')
plt.show()
```

## Interpretation of Correlation Analysis
The correlation heatmap demonstrates generally strong positive relationships between most youth age groups. This suggests that sports with strong participation among younger children often continue to maintain participation as children move into older age categories. In particular, the correlation between the 5–8 and 9–11 age groups appears especially strong, indicating that participation during middle childhood may strongly influence continued engagement during later childhood years.

However, some variation is still visible between the youngest age group (0–4) and older adolescent groups (12–14). This may suggest that certain sports experience participation changes as children become older and potentially face increasing academic pressures, social changes, or competing recreational interests.

From a Brisbane 2032 perspective, sports demonstrating stronger consistency across age groups may provide more stable long-term athlete development pathways and stronger opportunities for sustained Olympic fan engagement over time.

## K-Means Clustering: 
While correlation analysis helps identify relationships between age groups, clustering analysis allows sports to be grouped according to similar participation pathway patterns. Machine learning techniques such as K-Means clustering help uncover hidden structures within the data that may not be immediately visible through simple visualisation alone.

Clustering may help identify:

- sports with consistently strong participation
- sports with declining engagement during adolescence
- sports with late participation growth
- sports with unstable participation patterns.

These clusters may provide valuable strategic insight for future Olympic engagement planning and youth retention strategies.



```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X = age_data[age_columns]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

age_data['Cluster'] = kmeans.fit_predict(X_scaled)

age_data
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=age_data,
    x='5-8',
    y='12-14',
    hue='Cluster',
    s=150
)

for i in range(len(age_data)):
    plt.text(
        age_data['5-8'].iloc[i],
        age_data['12-14'].iloc[i],
        age_data['Activity'].iloc[i],
        fontsize=9
    )

plt.title('Olympic Sport Participation Pathway Clusters')
plt.xlabel('Participation Rate (Age 5-8)')
plt.ylabel('Participation Rate (Age 12-14)')
plt.show()
```

## Interpretation of Clustering Results: 
The clustering analysis identifies several distinct participation pathway patterns across Olympic sports. Some sports appear within clusters demonstrating relatively stable participation across both younger and older age groups, while other sports appear within clusters showing stronger participation during childhood but lower participation during adolescence.

Sports positioned within stronger long-term participation clusters may represent more sustainable athlete development pathways leading into Brisbane 2032. These sports may already possess stronger junior club systems, community engagement structures, and long-term participation culture.

Other sports demonstrating lower adolescent retention may require targeted intervention strategies designed to maintain engagement as children grow older. This may include:

improved youth development programs,

stronger community competition pathways,
and greater media visibility targeting adolescent audiences.

The clustering analysis also demonstrates how machine learning techniques can support sports strategy development by identifying hidden participation patterns that may not be immediately visible through traditional descriptive analysis alone.

These include article headlines, summary text, publication dates, and article sections.

Selecting only relevant attributes improves readability and creates a cleaner semi structured dataset for later text analysis. Combining headlines and summaries into a single text field also strengthens the quality of later preprocessing and keyword analysis because it captures a broader representation of article content.

```python
articles = data['response']['results']

guardian_data = []

for article in articles:
    headline = article['webTitle']
    summary = article.get('fields', {}).get('trailText', '')
    date = article['webPublicationDate']
    section = article['sectionName']
    combined_text = headline + ' ' + summary
    
    guardian_data.append({
        'Headline': headline,
        'Summary': summary,
        'Date': date,
        'Section': section,
        'Text': combined_text
    })

guardian_df = pd.DataFrame(guardian_data)
guardian_df.head()
```

The extracted dataset contains Olympic related media headlines and summaries connected to Australian sporting discussion. Combining article headlines and summaries improves the quality of later text analysis because it captures richer contextual information and broader language patterns.

## Cleaning and Processing Semi Structured Data:
Article text cannot be analysed effectively in its original form because news content often contains punctuation, stopwords, HTML formatting, and inconsistent text structures. Text preprocessing is therefore performed to standardise the dataset before applying analytical techniques.

The preprocessing stage converts all text into lowercase format, removes punctuation and HTML tags, tokenises the text into individual words, and removes common stopwords that do not contribute meaningful analytical value. These cleaning steps improve the reliability of keyword frequency analysis, sentiment analysis, and TF IDF calculations by reducing unnecessary variation within the text.

```python
import nltk
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub(r'<.*?>', '', text)
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    cleaned = ' '.join(tokens)
    
    return cleaned
```

```python
from nltk.tokenize import word_tokenize
```

```python
guardian_df['Cleaned_Text'] = guardian_df['Text'].apply(clean_text)
guardian_df[['Headline', 'Cleaned_Text']].head()
```

## Sentiment Analysis:
Sentiment analysis was conducted to examine the emotional tone of media coverage related to Olympic sports. Understanding media sentiment is important because positive or negative reporting may influence public perception, athlete visibility, youth inspiration, and long term engagement with sport. Within the Brisbane 2032 context, positive media representation may help strengthen emotional connection between younger audiences and Olympic sports. The VADER sentiment analysis model was applied to the cleaned article text to calculate compound sentiment scores for each article headline and summary. These scores were then classified into positive, neutral, and negative categories to allow clearer interpretation of overall media tone.

```python
from nltk.sentiment import SentimentIntensityAnalyzer
```

```python
sia = SentimentIntensityAnalyzer()

guardian_df['Sentiment_Score'] = guardian_df['Cleaned_Text'].apply(
    lambda text: sia.polarity_scores(text)['compound']
)

guardian_df[['Headline', 'Sentiment_Score']].head()
```

```python
def sentiment_label(score):
    if score > 0.05:
        return 'Positive'
    
    elif score < -0.05:
        return 'Negative'
    
    else:
        return 'Neutral'

guardian_df['Sentiment_Label'] = guardian_df['Sentiment_Score'].apply(sentiment_label)
guardian_df['Sentiment_Label'].value_counts()
```

```python
plt.figure(figsize=(8,5))

guardian_df['Sentiment_Label'].value_counts().plot(
    kind='bar'
)

plt.title('Sentiment Distribution of Olympic Sport Media Coverage')
plt.xlabel('Sentiment Category')
plt.ylabel('Number of Articles')
plt.show()
```

## Interpretation of Sentiment Analysis: 
The sentiment analysis demonstrates that most Olympic sport articles are likely to contain neutral or positive sentiment. Positive sentiment commonly appears in articles discussing:

- Athlete success
- Tournament victories

- Sporting achievement
- Youth inspiration

This is important for the Brisbane 2032 context because positive media narratives may contribute toward stronger emotional engagement with Olympic sports among younger audiences. Positive athlete representation may also help increase inspiration, role model visibility, and long-term sporting interest. Negative sentiment may appear in discussions involving injuries, controversies, or organisational issues, although these articles are typically less dominant overall.


## TF IDF Analysis:
TF IDF analysis was conducted to identify the most important and distinctive words appearing throughout the Olympic sport media dataset. Unlike simple frequency analysis, TF IDF reduces the influence of extremely common words and highlights terms that carry stronger analytical importance within the collection of articles. This provides deeper insight into the major themes, sporting discussions, and Olympic related topics receiving attention within digital media coverage.

Analysing these important terms is valuable because media discussion may influence public awareness, athlete visibility, and youth engagement with Olympic sport before the Brisbane 2032 Olympic Games. The analysis also helps identify whether media attention is focused mainly on elite competition or broader participation related themes.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(max_features=20)
tfidf_matrix = tfidf.fit_transform(guardian_df['Cleaned_Text'])
feature_names = tfidf.get_feature_names_out()
scores = tfidf_matrix.sum(axis=0).A1
```

```python
tfidf_df = pd.DataFrame({
    'Word': feature_names,
    'Score': scores
})
```

```python
tfidf_df = tfidf_df.sort_values(
    by='Score',
    ascending=False
)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=tfidf_df,
    x='Score',
    y='Word'
)

plt.title('Top TF-IDF Terms in Olympic Sport Media Coverage')
plt.xlabel('TF-IDF Score')
plt.ylabel('Keyword')
plt.show()
```

## Interpretation:
The TF IDF analysis highlights several highly important terms associated with Olympic sport media coverage. Words such as olympics, athletics, games, athletes, football, basketball, and gold receive relatively high TF IDF scores, indicating that these topics appear consistently across the collected articles while also carrying strong analytical importance. The results suggest that media reporting is strongly focused on elite sporting competition, international events, athlete performance, and tournament related discussion. Terms associated with Olympic achievement and professional sport appear more prominently than terms connected to community participation or grassroots engagement.

This finding is important within the Brisbane 2032 context because strong media visibility does not always translate directly into stronger youth participation outcomes. While elite sporting coverage may inspire some younger audiences, highly performance focused reporting may also create a perception that Olympic sport is primarily associated with professional level competition rather than accessible community participation.

The analysis therefore suggests that future engagement strategies may benefit from increasing youth focused storytelling, grassroots participation coverage, and athlete development narratives alongside elite competition reporting. Expanding media representation beyond professional competition may help strengthen broader youth connection with Olympic sport before Brisbane 2032.
